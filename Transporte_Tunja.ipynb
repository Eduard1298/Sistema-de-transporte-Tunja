```python
# ============================================================
# SISTEMA INTELIGENTE DE TRANSPORTE URBANO DE TUNJA
# ============================================================
#
# Actividad 2 - Búsqueda y sistemas basados en reglas
#
# Este programa utiliza una base de conocimiento con algunas
# rutas del transporte urbano de Tunja.
#
# El usuario indica un lugar de origen y un lugar de destino.
# El sistema busca la MEJOR ruta (la de menor tiempo) que
# permita llegar al destino.
#
# El programa utiliza:
# - Base de conocimiento
# - Reglas lógicas
# - Condicionales if, elif y else
# - Funciones
# - return
# - Listas y diccionarios
#
# --------------------------------------------------------
# MEJORAS AGREGADAS RESPECTO A LA VERSIÓN ANTERIOR:
# --------------------------------------------------------
# 1) Se corrigió un error en la búsqueda con transbordo: ahora
#    se verifica que el punto de transbordo realmente esté
#    ANTES del destino dentro de la segunda ruta.
# 2) El sistema ya no se queda con la primera ruta que
#    encuentra: revisa todas las posibles y escoge la de
#    menor tiempo (la "mejor" ruta).
# 3) Las rutas ahora se pueden recorrer en los dos sentidos
#    (de ida y de regreso), como ocurre con un bus real.
# 4) Se agregó validación: si el usuario escribe un lugar que
#    no existe en la base de conocimiento, el sistema se lo
#    informa en vez de simplemente decir "no se encontró ruta".
# ============================================================


# ============================================================
# 1. BASE DE CONOCIMIENTO
# ============================================================

# En esta lista almacenamos los lugares que nuestro sistema
# conoce y puede utilizar para buscar una ruta.

LUGARES = [
    "Arboleda",
    "Avenida Norte",
    "Terminal",
    "Cooservicios",
    "La Fuente",
    "Plaza de Bolívar",
    "ESE Santiago de Tunja",
    "Muiscas",
    "Triunfo",
    "Runta",
    "Avenida Oriental",
    "Nieves",
    "Uniboyacá",
    "Pirgua",
    "Estadio",
    "Retén Sur",
    "Asís",
    "Avenida Universitaria",
    "C.C. Viva",
    "San Francisco",
    "Florida"
]


# ============================================================
# 2. BASE DE CONOCIMIENTO DE LAS RUTAS
# ============================================================

# En este diccionario almacenamos las rutas.
#
# Cada ruta tiene una lista de lugares por los cuales pasa,
# en el orden en que un bus los recorre de ida. El sistema se
# encarga de revisar también el sentido contrario (de vuelta).
#
# Los recorridos son una representación simplificada para
# realizar la actividad académica.

RUTAS = {

    "R1": [
        "Arboleda",
        "Avenida Norte",
        "Terminal"
    ],

    "R8": [
        "Cooservicios",
        "La Fuente",
        "Plaza de Bolívar",
        "Terminal",
        "ESE Santiago de Tunja"
    ],

    "R9": [
        "Arboleda",
        "Avenida Norte",
        "La Fuente",
        "Avenida Universitaria"
    ],

    "R10": [
        "Muiscas",
        "Avenida Norte",
        "Triunfo"
    ],

    "R12": [
        "Runta",
        "Avenida Oriental",
        "Nieves",
        "Avenida Norte",
        "Muiscas",
        "Uniboyacá",
        "Plaza de Bolívar"
    ],

    "R15": [
        "Cooservicios",
        "Avenida Norte",
        "Muiscas"
    ],

    "R20": [
        "Pirgua",
        "Terminal",
        "Avenida Norte",
        "Arboleda"
    ],

    "R21": [
        "Pirgua",
        "Terminal",
        "Estadio",
        "Retén Sur"
    ],

    "R22": [
        "Asís",
        "Muiscas",
        "Avenida Universitaria",
        "C.C. Viva",
        "Terminal",
        "La Fuente"
    ],

    "R23": [
        "Arboleda",
        "Uniboyacá",
        "Florida",
        "San Francisco"
    ]
}


# ============================================================
# 3. TIEMPO ESTIMADO DE CADA RUTA
# ============================================================

# En esta parte almacenamos un tiempo aproximado para cada
# recorrido completo utilizado por nuestro sistema.
#
# Estos valores son utilizados únicamente para la simulación
# académica y no representan tiempos reales de viaje.

TIEMPOS = {

    "R1": 15,
    "R8": 20,
    "R9": 20,
    "R10": 15,
    "R12": 30,
    "R15": 20,
    "R20": 20,
    "R21": 20,
    "R22": 25,
    "R23": 25
}

# Tiempo adicional (en minutos) que se penaliza por cada
# transbordo que el usuario deba realizar.
TIEMPO_TRANSBORDO = 5


# ============================================================
# 4. FUNCIÓN PARA BUSCAR LA MEJOR RUTA DIRECTA
# ============================================================

def buscar_ruta_directa(origen, destino):

    # Aquí guardaremos la mejor ruta directa encontrada
    # hasta el momento (la de menor tiempo).
    mejor_ruta = None
    mejor_tiempo = None

    # Recorremos todas las rutas almacenadas en la base
    # de conocimiento.
    for nombre_ruta, lugares in RUTAS.items():

        # Revisamos el recorrido en su orden original (ida)
        # y también en orden invertido (regreso), ya que un
        # bus normalmente circula en los dos sentidos.
        sentido_ida = lugares
        sentido_regreso = list(reversed(lugares))

        for recorrido_completo in [sentido_ida, sentido_regreso]:

            # Comprobamos si el origen y el destino se
            # encuentran dentro de este recorrido.
            if origen in recorrido_completo and destino in recorrido_completo:

                # Obtenemos la posición del origen.
                posicion_origen = recorrido_completo.index(origen)

                # Obtenemos la posición del destino.
                posicion_destino = recorrido_completo.index(destino)

                # Comprobamos que el destino aparezca después
                # del origen dentro del recorrido.
                if posicion_origen < posicion_destino:

                    # Obtenemos el tiempo de esta ruta.
                    tiempo = TIEMPOS[nombre_ruta]

                    # Si es la primera ruta válida que encontramos,
                    # o si es más rápida que la mejor guardada hasta
                    # ahora, la convertimos en la nueva mejor opción.
                    if mejor_tiempo is None or tiempo < mejor_tiempo:

                        # Obtenemos solamente la parte del recorrido
                        # que necesita realizar el usuario.
                        recorrido = recorrido_completo[
                            posicion_origen:posicion_destino + 1
                        ]

                        mejor_tiempo = tiempo
                        mejor_ruta = (nombre_ruta, recorrido, tiempo)

    # Retornamos la mejor ruta directa encontrada (o None si
    # no existe ninguna).
    return mejor_ruta


# ============================================================
# 5. FUNCIÓN PARA BUSCAR LA MEJOR RUTA CON TRANSBORDO
# ============================================================

def buscar_ruta_con_transbordo(origen, destino):

    # Aquí guardaremos la mejor combinación de rutas
    # encontrada hasta el momento (la de menor tiempo total).
    mejor_opcion = None
    mejor_tiempo_total = None

    # Recorremos todas las rutas para usarlas como "primera ruta".
    for ruta_1, lugares_1_original in RUTAS.items():

        # Revisamos la primera ruta en sus dos sentidos posibles.
        for lugares_1 in [lugares_1_original, list(reversed(lugares_1_original))]:

            # Verificamos que el origen esté en esta ruta.
            if origen in lugares_1:

                posicion_origen = lugares_1.index(origen)

                # Recorremos nuevamente las rutas para usarlas
                # como "segunda ruta".
                for ruta_2, lugares_2_original in RUTAS.items():

                    # Evitamos comparar una ruta consigo misma.
                    if ruta_1 != ruta_2:

                        # Revisamos la segunda ruta en sus dos
                        # sentidos posibles.
                        for lugares_2 in [lugares_2_original, list(reversed(lugares_2_original))]:

                            # Verificamos si la segunda ruta
                            # contiene el destino.
                            if destino in lugares_2:

                                posicion_destino = lugares_2.index(destino)

                                # Buscamos un lugar donde ambas rutas
                                # puedan conectarse.
                                for lugar in lugares_1:

                                    if lugar in lugares_2:

                                        # Posición del punto de
                                        # transbordo en cada ruta.
                                        posicion_transbordo_1 = lugares_1.index(lugar)
                                        posicion_transbordo_2 = lugares_2.index(lugar)

                                        # Regla clave: el transbordo debe
                                        # quedar DESPUÉS del origen en la
                                        # primera ruta, y ANTES (o en el
                                        # mismo punto) del destino en la
                                        # segunda ruta. Esto es lo que
                                        # corrige el error de la versión
                                        # anterior.
                                        if posicion_origen < posicion_transbordo_1 and posicion_transbordo_2 <= posicion_destino:

                                            # Calculamos el tiempo total
                                            # de esta combinación de rutas.
                                            tiempo_total = (
                                                TIEMPOS[ruta_1] +
                                                TIEMPOS[ruta_2] +
                                                TIEMPO_TRANSBORDO
                                            )

                                            # Si es la primera combinación
                                            # válida, o si es más rápida
                                            # que la mejor guardada hasta
                                            # ahora, la convertimos en la
                                            # nueva mejor opción.
                                            if mejor_tiempo_total is None or tiempo_total < mejor_tiempo_total:

                                                # Construimos el primer tramo.
                                                primer_tramo = lugares_1[
                                                    posicion_origen:
                                                    posicion_transbordo_1 + 1
                                                ]

                                                # Construimos el segundo tramo,
                                                # comenzando justo en el punto
                                                # de transbordo.
                                                segundo_tramo = lugares_2[
                                                    posicion_transbordo_2:
                                                    posicion_destino + 1
                                                ]

                                                mejor_tiempo_total = tiempo_total

                                                mejor_opcion = (
                                                    ruta_1,
                                                    ruta_2,
                                                    primer_tramo,
                                                    segundo_tramo,
                                                    lugar,
                                                    tiempo_total
                                                )

    # Retornamos la mejor combinación encontrada (o None si
    # no existe ninguna).
    return mejor_opcion


# ============================================================
# 6. FUNCIÓN PARA VALIDAR QUE UN LUGAR EXISTA EN LA BASE
# ============================================================

def lugar_valido(lugar):

    # Esta función simplemente comprueba si el lugar escrito
    # por el usuario existe dentro de nuestra base de
    # conocimiento (la lista LUGARES).
    if lugar in LUGARES:
        return True
    else:
        return False


# ============================================================
# 7. FUNCIÓN PRINCIPAL DEL SISTEMA
# ============================================================

def sistema_transporte():

    # Mostramos el título del programa.
    print("\n==============================================")
    print(" SISTEMA INTELIGENTE DE TRANSPORTE DE TUNJA")
    print("==============================================")

    # Mostramos los lugares que conoce el sistema.
    print("\nLugares disponibles:")

    for lugar in LUGARES:
        print("-", lugar)

    # Solicitamos al usuario el lugar de origen.
    origen = input("\nIngrese el origen: ")

    # Solicitamos al usuario el lugar de destino.
    destino = input("Ingrese el destino: ")


    # --------------------------------------------------------
    # REGLA 0
    # --------------------------------------------------------
    #
    # Antes de buscar cualquier ruta, comprobamos que los
    # lugares escritos existan en la base de conocimiento.

    if not lugar_valido(origen) or not lugar_valido(destino):

        print("\n==============================================")
        print(" LUGAR NO RECONOCIDO")
        print("==============================================")

        print(
            "\nUno de los lugares que escribiste no está "
            "registrado en la base de conocimiento."
        )
        print(
            "Verifica que el nombre esté escrito exactamente "
            "igual a como aparece en la lista de lugares "
            "disponibles."
        )

        # Terminamos la función, ya que no tiene sentido
        # seguir buscando una ruta.
        return


    # --------------------------------------------------------
    # REGLA 1
    # --------------------------------------------------------
    #
    # Si el origen y el destino están en una misma ruta,
    # entonces utilizamos la ruta directa más rápida.

    resultado_directo = buscar_ruta_directa(
        origen,
        destino
    )


    # --------------------------------------------------------
    # REGLA 2
    # --------------------------------------------------------
    #
    # Buscamos también la mejor combinación con transbordo,
    # para poder compararla con la ruta directa (si existe)
    # y así ofrecer siempre la opción más rápida.

    resultado_transbordo = buscar_ruta_con_transbordo(
        origen,
        destino
    )


    # --------------------------------------------------------
    # COMPARACIÓN: elegimos la mejor opción disponible
    # --------------------------------------------------------

    # Caso 1: existen las dos opciones, comparamos tiempos.
    if resultado_directo and resultado_transbordo:

        nombre_ruta, recorrido, tiempo_directo = resultado_directo
        tiempo_transbordo = resultado_transbordo[5]

        if tiempo_directo <= tiempo_transbordo:
            usar_directa = True
        else:
            usar_directa = False

    # Caso 2: solo existe la ruta directa.
    elif resultado_directo and not resultado_transbordo:
        usar_directa = True

    # Caso 3: solo existe la ruta con transbordo.
    elif resultado_transbordo and not resultado_directo:
        usar_directa = False

    # Caso 4: no existe ninguna de las dos.
    else:

        print("\n==============================================")
        print(" NO SE ENCONTRÓ UNA RUTA")
        print("==============================================")

        print(
            "\nEl sistema no encontró una conexión "
            "registrada entre los lugares seleccionados."
        )
        print(
            "Intente nuevamente utilizando "
            "lugares disponibles en la base de conocimiento."
        )

        # Terminamos la función.
        return


    # --------------------------------------------------------
    # MOSTRAR LA RUTA DIRECTA (si fue la elegida)
    # --------------------------------------------------------

    if usar_directa:

        nombre_ruta, recorrido, tiempo = resultado_directo

        print("\n==============================================")
        print(" RUTA DIRECTA ENCONTRADA (la más rápida)")
        print("==============================================")

        print("\nRuta recomendada:", nombre_ruta)

        print("\nRecorrido:")

        for lugar in recorrido:
            print("→", lugar)

        print("\nTiempo aproximado:", tiempo, "minutos")

        print("Transbordos: 0")

        print("\nLa ruta fue encontrada mediante una regla directa.")


    # --------------------------------------------------------
    # MOSTRAR LA RUTA CON TRANSBORDO (si fue la elegida)
    # --------------------------------------------------------

    else:

        (
            ruta_1,
            ruta_2,
            primer_tramo,
            segundo_tramo,
            transbordo,
            tiempo_total
        ) = resultado_transbordo

        print("\n==============================================")
        print(" RUTA CON TRANSBORDO ENCONTRADA (la más rápida)")
        print("==============================================")

        print("\nPrimera ruta:", ruta_1)

        print("Primer recorrido:")

        for lugar in primer_tramo:
            print("→", lugar)

        print("\nDebe realizar un transbordo en:", transbordo)

        print("\nSegunda ruta:", ruta_2)

        print("Segundo recorrido:")

        for lugar in segundo_tramo:
            print("→", lugar)

        print("\nTransbordos: 1")

        print(
            "\nTiempo aproximado:",
            tiempo_total,
            "minutos"
        )

        print(
            "\nLa ruta fue encontrada mediante "
            "las reglas de conexión."
        )


# ============================================================
# 8. EJECUTAR EL SISTEMA
# ============================================================

# Esta instrucción inicia nuestro sistema inteligente.

sistema_transporte()
```
